## 1. Importación de librerías

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 12

print("Librerías cargadas correctamente.")
```

## 2. Función auxiliar para calcular Pearson paso a paso

Esta función muestra los mismos elementos que calculas manualmente en el cuaderno.

```python
def pearson_manual_detalle(x, y, nombre_x="X", nombre_y="Y"):
    """
    Calcula el coeficiente de Pearson mostrando todos los pasos
    (equivalente al cálculo manual).
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    n = len(x)
    
    sum_x = np.sum(x)
    sum_y = np.sum(y)
    sum_x2 = np.sum(x**2)
    sum_y2 = np.sum(y**2)
    sum_xy = np.sum(x * y)
    
    numerador = n * sum_xy - sum_x * sum_y
    denom_x = n * sum_x2 - sum_x**2
    denom_y = n * sum_y2 - sum_y**2
    denominador = np.sqrt(denom_x * denom_y)
    
    r = numerador / denominador
    
    # También con scipy para verificar y obtener p-valor
    r_scipy, p_valor = stats.pearsonr(x, y)
    
    print("=" * 65)
    print(f"Cálculo de Pearson entre {nombre_x} y {nombre_y}")
    print("=" * 65)
    print(f"n          = {n}")
    print(f"Σx         = {sum_x:.4f}")
    print(f"Σy         = {sum_y:.4f}")
    print(f"Σx²        = {sum_x2:.4f}")
    print(f"Σy²        = {sum_y2:.4f}")
    print(f"Σxy        = {sum_xy:.4f}")
    print("-" * 65)
    print(f"Numerador  = {numerador:.4f}")
    print(f"Denominador= {denominador:.4f}")
    print(f"r (manual) = {r:.4f}")
    print(f"r (scipy)  = {r_scipy:.4f}")
    print(f"p-valor    = {p_valor:.6e}")
    print("=" * 65)
    
    return r, p_valor
```

## 3. Verificación del Ejercicio 1 (cuaderno)

Datos del **Ejercicio 1** de los ejercicios manuales:

| x (horas) | y (calificación) |
|-----------|------------------|
| 1         | 55               |
| 2         | 60               |
| 4         | 75               |

| 7         | 90               |

Compara el resultado que obtuviste a mano con el que genera el código.

```python
x1 = [1, 2, 4, 5, 7]
y1 = [55, 60, 75, 80, 90]

r1, p1 = pearson_manual_detalle(x1, y1, "Horas de estudio", "Calificación")

# Visualización
plt.figure(figsize=(8, 5))
sns.regplot(x=x1, y=y1, ci=None, scatter_kws={"s": 100, "color": "steelblue"},
            line_kws={"color": "red", "linewidth": 2})
plt.title(f"Ejercicio 1: r = {r1:.4f}")
plt.xlabel("Horas de estudio")
plt.ylabel("Calificación")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 5. Efecto de un outlier (Ejercicio 3)

Comparamos el coeficiente **con** y **sin** el valor atípico.

```python
# Datos completos (con outlier)
x3_completo = [8, 10, 12, 15, 18, 50]
y3_completo = [1.2, 1.5, 1.8, 2.1, 2.4, 9.0]

# Datos sin outlier
x3_sin = [8, 10, 12, 15, 18]
y3_sin = [1.2, 1.5, 1.8, 2.1, 2.4]

print(">>> CON el outlier (x=50, y=9.0):")
r3_con, _ = pearson_manual_detalle(x3_completo, y3_completo, "Ingreso", "Gasto")

print("\n>>> SIN el outlier:")
r3_sin, _ = pearson_manual_detalle(x3_sin, y3_sin, "Ingreso", "Gasto")

# Visualización comparativa
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.regplot(x=x3_completo, y=y3_completo, ax=axes[0], ci=None,
            scatter_kws={"s": 100, "color": "crimson"},
            line_kws={"color": "black", "linewidth": 2})
axes[0].set_title(f"Con outlier\nr = {r3_con:.4f}")
axes[0].set_xlabel("Ingreso (miles)")
axes[0].set_ylabel("Gasto entretenimiento (miles)")

sns.regplot(x=x3_sin, y=y3_sin, ax=axes[1], ci=None,
            scatter_kws={"s": 100, "color": "seagreen"},
            line_kws={"color": "black", "linewidth": 2})
axes[1].set_title(f"Sin outlier\nr = {r3_sin:.4f}")
axes[1].set_xlabel("Ingreso (miles)")
axes[1].set_ylabel("Gasto entretenimiento (miles)")

plt.tight_layout()
plt.show()

print(f"\nDiferencia en r: {abs(r3_con - r3_sin):.4f}")
print("Observa cómo un solo punto cambia drásticamente el coeficiente.")
```

## 6. Aplicación a un dataset real: `tips` (seaborn)

Usaremos el dataset clásico de propinas para practicar el cálculo e interpretación en un contexto real.

```python
tips = sns.load_dataset("tips")
print("Dimensiones:", tips.shape)
display(tips.head(8))
print("\nVariables numéricas:")
display(tips.describe().round(2))
```

```python
# Correlación entre total_bill y tip
r_tips, p_tips = pearson_manual_detalle(
    tips["total_bill"], tips["tip"],
    "Total de la cuenta", "Propina"
)

# Visualización
plt.figure(figsize=(9, 6))
sns.regplot(data=tips, x="total_bill", y="tip",
            scatter_kws={"alpha": 0.6, "s": 50},
            line_kws={"color": "red", "linewidth": 2})
plt.title(f"Total de la cuenta vs Propina\nr = {r_tips:.4f}, p = {p_tips:.2e}")
plt.xlabel("Total de la cuenta ($)")
plt.ylabel("Propina ($)")
plt.tight_layout()
plt.show()
```

## 7. Matriz de correlación y mapa de calor

```python
# Solo variables numéricas
num_cols = tips.select_dtypes(include="number")
corr_matrix = num_cols.corr(method="pearson")

print("Matriz de correlación de Pearson:")
display(corr_matrix.round(3))

plt.figure(figsize=(7, 5))
sns.heatmap(corr_matrix, annot=True, cmap="RdBu_r", center=0,
            vmin=-1, vmax=1, square=True, linewidths=0.5,
            fmt=".3f")
plt.title("Mapa de calor – Coeficiente de Pearson (tips)")
plt.tight_layout()
plt.show()
```
